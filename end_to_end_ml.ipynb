```python
import pandas as pd
import numpy as np
```

**Step 1 (Preprocessing)**

remove the column from at index '0'

```python
df = df.iloc[:,1:] # ':'--> means select all rows ',1:' --> means starting from first column to further till end rather then 0th column
df
```

```python
df.shape
```

**Step 2 --> EDA**

For EDA we make a plot using matplotlib libarary

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df["CGPA"],df["IQ"],c=df["Placement"])
plt.title("Scatter Plot")
plt.xlabel("CGPA")
plt.ylabel("IQ")
plt.show()
#
```

**Step 3 --> test train split**

```python
X = df.iloc[:,0:2]  # [rows,columns]
y = df.iloc[:,-1]
```

```python
X
```

```python
y
```

```python
from os import X_OK
# model load
from sklearn.model_selection import train_test_split
X_train,X_test, y_train,y_test  = train_test_split(X,y,test_size=0.1,random_state=2)
```

```python
X_train
```

```python
y_train
```

```python
X_test
```

```python
y_test
```

**Step 4 --> Scalling**

```python
from sklearn.preprocessing import StandardScaler
scaller = StandardScaler() #standard scaller range from (-1 to +1)
X_train = scaller.fit_transform(X_train)
```

```python
X_train
```

```python
X_test = scaller.transform(X_test)
```

```python
X_test
```

**Step 5 Model Training**

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

**Step 6 Evaluate Model**

```python
from sklearn.metrics import accuracy_score
```

```python
y_pred = model.predict(X_test)
```

```python
y_test
```

```python
accuracy_score(y_test,y_pred)
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(X_train, y_train.values, clf=model, legend=2)
```
